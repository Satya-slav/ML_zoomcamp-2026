q6 median value of horsepower

q7 sum of weights

```python
import numpy as np
asia_df = df[df['origin'] == 'Asia'][['engine_displacement', 'horsepower', 'vehicle_weight']].dropna()
X = asia_df.iloc[:9].values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 800, 750, 850, 1300, 1000, 1000, 1300, 800])
w = XTX_inv @ X.T @ y
print("Sum of weights:", w.sum())
```
